```python
import json
```

```python
path = "../data/dev_notes.jsonl"
```

```python
with open(path, "r") as f:
    data = [json.loads(line) for line in f]
```

# DATA EXPLORATION

This is to understand the development data - The below results show that the golden data can have more than one IDC-11 code

```python
for item in data:
    if len(item["gold"]) > 1:
        print(item["gold"])
```

```python
dup_dict = {}
for item in data:
    id = item["id"]
    if id in dup_dict:
        dup_dict[id] += 1
    else:
        dup_dict[id] = 1

for k,v in dup_dict.items():
    if v >1 :
        print(k)
```

```python
style_dict = {}
for item in data:
    style = item["style"]
    if style in style_dict:
        style_dict[style] += 1
    else:
        style_dict[style] = 1

for k,v in style_dict.items():
    print(k, v)
```

The data contains three types of text:

1. **`short_consult`**: Contains a brief summary of the consultation, including the patient's symptoms, prescribed medications, age, and other relevant details.

2. **`typed_consult`**: Contains a more detailed description of the consultation, including the patient's symptoms, medical history, lifestyle, occupation, and other relevant information.

3. **`ambient_transcript`**: Contains the conversation transcript between the doctor and the patient, capturing the details discussed during the consultation.

```python
style_dict = {}
for item in data:
    style = item["style"]
    if style in style_dict:
        style_dict[style].append(item["id"])
    else:
        style_dict[style] = [item["id"]]

for k,v in style_dict.items():
    print(k, v)
```
